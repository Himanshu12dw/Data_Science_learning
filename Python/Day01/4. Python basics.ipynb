## 3. Variables

A variable is a container used to store data in memory.

Instead of using actual values repeatedly, we store them in variables and use the variable name throughout the program.

#### Syntax

```python
variable_name = value
```

#### Why Use Variables?

- Store data for later use
- Make code readable
- Avoid repeating values
- Easily update values

#### Variable Naming Rules

##### Valid

```python
name = "Himanshu"
age = 25
student_name = "Rahul"
salary_2025 = 50000
```

##### Invalid

```python
1name = "Himanshu"
student-name = "Rahul"
class = "Python"
my name ="himanshu"
```

#### Naming Convention

Use snake_case:

```python
student_name = "Himanshu"
total_salary = 50000
```

#### Key Points

- Variables are case-sensitive.
- Python does not require data type declaration.
- Variable values can be changed during execution.

````python
name = "Himanshu"     #python is a dynamically typed language. We do not need to declare the data type of a variable.
age = 25

print(name)
print(age)
````

````python
salary = 50000

print(salary)

salary = 60000

print(salary)
````

````python
a = 10
b = 20

print(a + b)
````

## 5. Raw Strings

Normally Python interprets escape sequences.

A raw string tells Python to treat backslashes as normal characters.

### Syntax

```python
r"string"
```

### Use Cases

- File paths
- Regular Expressions

### Key Point

Raw strings ignore escape sequence processing.

````python
print("C:\newfolder\test")
# this code will give an error because \n is treated as a newline character
````

````python
# To avoid this, we can use raw strings by prefixing the string with 'r'
print(r"C:\newfolder\test")
````

## 7. round() Function

The `round()` function is used to round a number to the nearest integer or to a specified number of decimal places.

#### Syntax

```python
round(number, digits)
```

##### Parameters

- number → Value to round
- digits → Number of decimal places (optional)

#### Why Use round()?

- Reduce decimal places
- Improve readability
- Format numerical results

#### Key Points

- If digits are not provided, Python rounds to the nearest integer.
- If digits are provided, Python rounds to that many decimal places.

````python
print(round(3.7))
print(round(3.14159265, 2))

price = 199.98765
print(round(price, 2))
````

#### Identity Operators

Identity operators compare whether two variables refer to the same object in memory.

#### Identity Operators

| Operator | Meaning |
|----------|----------|
| is | Same object |
| is not | Different objects |

#### Important

Identity operators do NOT compare values.

They compare memory locations.

#### Value Comparison

```python
==
```

#### Memory Comparison

```python
is
```

### Key Point

`==` checks value.

`is` checks identity (memory location).

````python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)

# Values are same.
# Memory locations are different.
````

````python
a = [1, 2, 3]

b = a

print(a is b)

# Both variables point to the same object/memory location..
````

## 9. Operator Precedence (Python BODMAS)

When multiple operators are present in an expression, Python follows a priority order.

This is similar to BODMAS in mathematics.

#### Order of Execution

1. Parentheses `()`
2. Power `**`
3. Multiplication `*`
4. Division `/`
5. Floor Division `//`
6. Modulus `%`
7. Addition `+`
8. Subtraction `-`
9. Comparison Operators (`>, <, ==, !=`)
10. Logical `not`
11. Logical `and`
12. Logical `or`

---

#### Easy Formula

```text
()
**
* / // %
+ -
Comparison Operators
not
and
or
```

#### Key Point

When in doubt, use parentheses `()` to make calculations clear and avoid confusion.

````python
print(10 + 2 * 5)
# 2 * 5 = 10 ---> 10 + 10 = 20
````
