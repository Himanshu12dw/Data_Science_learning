## 4. String Slicing

Slicing extracts a portion of a string.

#### Syntax

```python
string[start:end:step]
```

Rule:

- Start included
- End excluded

````python
text = "Python"

print(text[:4])   #first 4 characters
print(text[2:4])  #from 3rd to 4th position
print(text[2:])   #from 3rd position to end of the string
print(text[:])    # whole string

# using step
print(text[0::2])   # every 2nd character starting from 1st position
print(text[4::-1])   # reverse the string from 5rd char   #start will be included
print(text[4:0:-1])   # reverse the string from 5rd char   #end will be excluded

#negetive step
print(text[5::-1])    # reverse string
````

## 6. String Methods

Methods belong to a string object.

Syntax:

```python
string.method()
```

Example:

```python
name.upper()
```

Methods are actions that strings can perform.

#### Difference

Function:

```python
len(name)
```

Method:

```python
name.upper()
```
#### Important String Methods

1. upper(): Converts all characters to uppercase.
2. lower(): Converts all characters to lowercase.
3. title(): Capitalizes the first letter of every word.
4. capitalize(): Capitalizes the first letter.
5. Strip(): Removes leading and trailing spaces.
6. replace(): Replaces one value with another.
7. split(): Splits a string into a list.
8. join(): Joins elements of a list into a string.
9. find(): Returns index of first occurrence.
10. count(): Counts occurrences of a substring.
11. swapcase(): to swap the cases



````python
first_name = "hiManshu"
middle_name = "   Kumar  " 
last_name = "DwiVEDi"
full_name = first_name + " " + middle_name.strip() + " " + last_name
print(full_name)

# 1. upper
print("use of upper method")
print(first_name.upper())
print()

# 2. lower
print("use of lower method")
print(last_name.lower())
print()

# 3. title
print("use of title method")
print(full_name.title())   
print()    

# 4. capitalize
print("use of capitalize method")
print(full_name.capitalize())
print()

# 5. swapcase
print("use of swapcase method")
print(first_name.swapcase())
print()
````

````python
# 6. count
print("use of count method")
print(first_name.count("h"))    # case sensitive
print()

# 7. find
print("use of find method")
print(first_name.find("a"))     # case sensitive, if not found it will return -1
print(first_name.lower().find("h",2))     # it will find the index of first occurrence of "h" starting from index 2, if not found it will return -1
print()

# 8. replace
print("use of replace method")
print(first_name.replace("h", "H"))   # case sensitive, it will replace all occurrences of "h" with "H"
print(last_name.replace("i", "E", 2))   # case sensitive, it will replace first 3 occurrences of "i" with "E"
print()


# 9. strip
print("use of strip method")
print(middle_name.strip())   # it will remove leading and trailing spaces
print()

# 10. split
print("use of split method")
print(full_name.split())    # it will split the string into a list of words based on whitespace
print()

# 11. join
print("use of join method")
print("-".join(full_name.split()))   # it will join the list of words with "-" as separator
print()


# 12. lstrip
print("use of lstrip method")
print(first_name+middle_name.lstrip()+last_name)   # it will remove leading spaces 
print()

# 13. rstrip
print("use of rstrip method")
print(first_name+middle_name.rstrip()+last_name)   # it will remove trailing spaces
print()
````

## 9. User Input

The `input()` function allows users to enter data while the program is running.

Without input, programs always work with hardcoded values.

With input, programs become interactive.

#### Syntax

```python
input("Prompt")
```

#### Important

The `input()` function always returns data as a string (`str`).

Even if the user enters a number, Python treats it as text. so change the data type if you want it to act like a number

#### Key Point

input() → Always returns a string.

````python
user_input = input("Enter your name: ")
user_age = input("Enter your age: ")
print("Hello " + user_input + ", you are " + user_age + " years old.")
````

````python
# more than two inputs can be taken by input function by using comma separation
name, age = input("Enter your name and age separated by comma: ").split(",")  #item assignment, it will split the input string into two parts based on comma and assign them to name and age variables respectively

print("Hello " + name + ", you are " + age + " years old.")
````

## 10 String Formatting

String Formatting is used to insert variables into strings in a clean and readable way.

Instead of manually joining strings, formatting creates dynamic text.

#### Why Is It Needed?

Imagine printing:

```text
My name is Himanshu and I am 25 years old.
```

The values may change every time.

String formatting allows variables to be inserted automatically.

#### Methods

1. Comma Separation
2. Concatenation
3. format()
4. f-Strings (Recommended)

#### Key Point

f-Strings are the modern and most commonly used formatting method.

````python
name = "Himanshu"
age = 25

# 1. Comma Separation 
print("My name is", name, "and I am", age, "years old.")
````

````python
# 2. String Formatting using concatenation
print("My name is " + name + " and I am " + str(age) + " years old.")
````

````python
# 3. String Formatting using format() method
print("My name is {} and I am {} years old.".format(name, age))
````

````python
# 4. String Formatting using f-string
print(f"My name is {name} and I am {age} years old.") 

#most recommended way to format string in python, it is more readable and efficient than other methods
````

````python
# expression inside f string
a = 10
b = 20

print(f"Sum = {a + b}")
````
