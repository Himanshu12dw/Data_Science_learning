## 15. Data Type

A data type defines what kind of data a value represents.

Common Data Types:

```python
int
float
str
bool
```

Data Type = Type of data being stored.

````python
age = 25
salary = 50000.5
name = "Himanshu"
is_active = True

print(type(age))        # Output: <class 'int'>
print(type(salary))     # Output: <class 'float'>
print(type(name))       # Output: <class 'str'>
print(type(is_active))  # Output: <class 'bool'>
````

## 16. Indentation



### Definition
Indentation refers to the spaces (or tabs) at the beginning of a line.

Python uses indentation to define code blocks instead of `{}`.

### Purpose
- Defines code blocks
- Improves readability
- Makes code structure clear

### Standard Rule
Use 4 spaces per indentation level.

### Colon and Indentation
A colon (`:`) is usually followed by an indented block.

Examples:
- if
- for
- while
- def
- class

### Common Errors
- Missing indentation
- Inconsistent indentation
- Mixing tabs and spaces

### Key Point
Indentation is part of Python syntax, not just formatting.

#### summary

- Indentation = spaces/tabs at the beginning of a line.
- Python uses indentation to create code blocks.
- A colon (`:`) is followed by an indented block.
- Standard indentation = 4 spaces.
- Indentation replaces `{}` used in many languages.
- Incorrect indentation causes IndentationError.

**Dedentation:-** Returning to a previous indentation level.

**IndentationError:** expected an indented block

### Identifier vs Variable

#### Identifier

An identifier is a **name** used to identify a programming element.

Examples of programming elements:

- Variables
- Functions
- Classes
- Modules
- Objects

#### Examples

```python
name
age
calculate_salary
Student
```

All of the above are identifiers because they are names.

---

#### Variable

A variable is a named memory location used to store data.

##### Example

```python
name = "Himanshu"
age = 25
```

Here:

- `name` is a variable
- `age` is a variable

Since variables have names, they are also identifiers.

---

##### Key Difference

- Identifier = Any valid name in Python.
- Variable = An identifier that stores data.

#### Formula

```text
Variable ⊂ Identifier
```

(Variable is a subset of Identifier)

````python
def greet():
    print("Hello")        #greet is not a variable but is an identier.
````
