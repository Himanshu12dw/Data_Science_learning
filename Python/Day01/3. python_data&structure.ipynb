## 2. Data Types

A Data Type defines what kind of data a value represents.

Python needs to know:

- Is it a number?
- Is it text?
- Is it True/False?
- Is it a collection?

#### Why Are Data Types Needed?

Different types of data support different operations.

Example:

```python
10 + 20
```

is valid because they are numbers.

```python
"Hello" + "World"
```

is valid because they are strings.

Python uses data types to decide what operations are allowed.

#### Key Point

Data Type = Classification of data.

### 2.1 Basic Data Types

| Data Type | Description | Example |
|------------|------------|------------|
| int | Integer | 10 |
| float | Decimal Number | 10.5 |
| str | Text/String | "Python" |
| bool | Boolean | True |

These are the most commonly used data types.

````python
age = 25
salary = 50000.75
name = "Himanshu"
is_active = True

print(type(age))
print(type(salary))
print(type(name))
print(type(is_active))
````

## 4. Data Structures


A Data Structure is a way of organizing and storing multiple pieces of data efficiently.

Basic data types store a single value.

Data structures store multiple values together.

#### Example

Single Value:

```python
age = 25
```

Multiple Values:

```python
ages = [20, 25, 30, 35]
```

#### Why Are Data Structures Needed?

Imagine storing marks of 100 students.

Without data structures:

```python
mark1 = 90
mark2 = 80
mark3 = 85
...
```

This becomes difficult.

Instead:

```python
marks = [90, 80, 85]
```

#### Key Point

Data Structure = Organized collection of data.

---
### 4.1 Python Data Structures

| Data Structure | Ordered | Mutable | Example |
|----------------|----------|----------|----------|
| List | Yes | Yes | [1, 2, 3] |
| Tuple | Yes | No | (1, 2, 3) |
| Set | No | Yes | {1, 2, 3} |
| Dictionary | Yes | Yes | {"name":"Himanshu"} |


These are the four primary Python data structures.

---
### 4.1.1 List

A List stores multiple items in order.

Lists are mutable, meaning they can be changed after creation.

#### Syntax

```python
[item1, item2, item3]
```

````python
skills = ["Python", "SQL", "Power BI"]

print(skills)
````

---
### 4.1.2 Tuple

A Tuple stores multiple items in order.

Unlike lists, tuples cannot be modified after creation.

#### Syntax

```python
(item1, item2, item3)
```

````python

coordinates = (10, 20, "name")

print(coordinates)
````

---
### 4.1.3 Sets

A Set stores unique values.

Duplicates are automatically removed.

#### Syntax

```python
{item1, item2, item3}
```

````python
numbers = {1, 2, 2, 3, 4, "you"}

print(numbers)
````

___
### 4.1.4 Dictionary

A Dictionary stores data as key-value pairs.

#### Syntax

```python
{
    key: value
}
```

Real-world example:

Name → Himanshu
Age → 25
City → Delhi

````python
student = {
    "name": "Himanshu",
    "age": 25,
    "city": "Delhi"
}

print(student)
````

---
### 4.2 Data Type vs Data Structure

| Data Type | Data Structure |
|------------|------------|
| Represents a type of data | Organizes multiple data values |
| Stores single value | Stores multiple values |
| Examples: int, float, str, bool | Examples: list, tuple, set, dict |

#### Example

Data Type:

```python
age = 25
```

Data Structure:

```python
ages = [20, 25, 30]
```

````python
age=25
ages=[23,34,52]

print(type(age), type(ages), sep=", ")
````

## 5. Data Type Conversion (Type Casting)

Data = Information

Data Type = Type of information

Data Structure = Organized collection of information

Type Casting means converting one data type into another.

#### Why Is It Needed?

The `input()` function returns strings.

If you want mathematical operations, you must convert strings into numbers.

#### Common Type Conversion Functions

| Function | Converts To |
|----------|------------|
| int() | Integer |
| float() | Decimal Number |
| str() | String |
| bool() | Boolean |

#### Key Point

Type Casting allows data to be used in different ways.

````python
age = 25

print("Age: " + str(age))
````

````python
#check data type
num = "100"

print(type(num))

num = int(num)

print(type(num))
````

````python
salary = "50000.50"
# salary = int(salary)   # we cannot convert float string to integer directly, it will give error
salary = int(float(salary)) # first we convert string to float and then float to integer, now it will work fine


print(salary)
````
