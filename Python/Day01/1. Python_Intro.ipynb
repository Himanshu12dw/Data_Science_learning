### What is python?

Python is a high-level, interpreted programming language used for software development, automation, web development, data analysis, machine learning, artificial intelligence, and scientific computing.


### Where Python is Used

- **Data Analysis** – analyzing and cleaning data
- **Data Science & AI/ML** – building predictive models
- **Web Development** – websites and APIs
- **Automation** – automating repetitive tasks
- **Cloud & DevOps** – managing cloud infrastructure
- **Cybersecurity** – scripting and security tools
- **Scientific Computing** – research and simulations

### Why Python

- Easy syntax
- Large community support
- Huge number of libraries
- Fast development

### How it works

Python executes code line by line through an interpreter. Instead of converting the entire code into machine code before execution, it translates and executes instructions during runtime.

# Python is:-

## 1. High-Level Language

A **high-level language** is a programming language that is easy for humans to read, write, and understand. It uses English-like syntax and hides complex hardware details from the programmer.

**Key Point:**
**High-Level = Human-friendly language**

### Benefits

* Easy to learn and write
* Easier to understand and maintain
* Focus on solving problems rather than hardware details

---

## 2. Interpreted Language

An **interpreted language** executes code line by line using an interpreter. Python reads and executes code one statement at a time.

**Key Point:**
**Interpreted = Code is executed through an interpreter instead of being fully compiled first**

### Advantages

* Easier debugging
* Faster development
* Platform independent

### Disadvantages

* Generally slower than compiled languages

---

## 3. Object-Oriented Language (OOP)

**Object-Oriented Programming (OOP)** organizes code using **Classes** and **Objects**.

* **Class:** Blueprint or template
* **Object:** Real instance created from a class

### Example

* Class → Car
* Objects → BMW, Audi, Tesla

### Why OOP is Useful

* Code reusability
* Easier maintenance
* Better code organization
* Scalable applications

**Key Point:**
**OOP = Programming using Classes and Objects**

---

## 4. Open Source

**Open Source** means the source code is publicly available.

Anyone can:

* View it
* Modify it
* Improve it
* Contribute to it

### Benefits

* Free to use
* Large community support
* Continuous improvements
* Thousands of free libraries

**Key Point:**
**Open Source = Source code is publicly available and free to use**

---

## 5. Cross-Platform

**Cross-platform** means the same code can run on different operating systems with little or no modification.

Python works on:

* Windows
* Linux
* macOS

This is possible because Python interpreters are available for multiple operating systems.

### Benefits

* Write once
* Run anywhere
* Easy deployment across systems

**Key Point:**
**Cross-Platform = Same code works across different operating systems**

---

# Summary

Python is:

1. **High-Level** → Easy for humans to read and write.
2. **Interpreted** → Executes code through an interpreter.
3. **Object-Oriented** → Uses Classes and Objects.
4. **Open Source** → Free and publicly available.
5. **Cross-Platform** → Runs on Windows, Linux, and macOS with minimal changes.

These features make Python one of the most popular programming languages for Data Science, Machine Learning, AI, and Software Development.
