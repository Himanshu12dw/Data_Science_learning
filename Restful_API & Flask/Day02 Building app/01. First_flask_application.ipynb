# 1. What is a Flask Application?

A Flask application is a Python program that receives HTTP requests, processes them, and returns responses.

Think of Flask as a bridge between:

```text
Browser → Flask → Python Code → Flask → Browser
```

Examples:

- Portfolio website
- Blog application
- Machine Learning API
- Dashboard application

---

# 2. Building a Basic Flask Application

## Minimal Flask App

```python
from flask import Flask

app = Flask(__name__)

@app.route('/')
def home():
    return "Hello World"

if __name__ == "__main__":
    app.run(debug=True)
```

### What Happens?

1. Flask application is created.
2. Route `/` is registered.
3. Browser visits `/`.
4. Function executes.
5. Response is returned.

# 5. Routing and URL Mapping

Routing is the process of connecting a URL to a Python function.

Example:

```python
@app.route('/')
def home():
    return "Home Page"
```

URL:

```text
http://localhost:5000/
```

calls:

```python
home()
```

---

# 6. Why Routing Exists

Without routing:

```text
Browser requests page
↓
Server doesn't know what function to execute
```

Routing solves this problem.

```text
URL
↓
Route
↓
Python Function
↓
Response
```

---

# 7. Static Routing

Static routes never change.

Example:

```python
@app.route('/')
def home():
    return "Home"

@app.route('/about')
def about():
    return "About"
```

Available URLs:

```text
/
about
```

Use Cases:

- Home page
- Contact page
- About page

---

# 8. Dynamic Routing

Dynamic routes accept variables.

Example:

```python
@app.route('/user/<name>')
def user(name):
    return f"Hello {name}"
```

Requests:

```text
/user/Himanshu
```

Output:

```text
Hello Himanshu
```

---

# 9. Dynamic Route Converters

## String

```python
@app.route('/user/<name>')
```

## Integer

```python
@app.route('/user/<int:id>')
```

Example:

```text
/user/10
```

id = 10

---

## Float

```python
@app.route('/price/<float:value>')
```

Example:

```text
/price/99.5
```

# 11. add_url_rule()

Alternative way to create routes.

Example:

```python
def home():
    return "Home Page"

app.add_url_rule(
    '/',
    'home',
    home
)
```

Useful when:

- Creating routes dynamically
- Reusing handlers
- Building large applications

---

# 12. Redirecting URLs

Sometimes a page moves.

Example:

```python
from flask import redirect

@app.route('/old')
def old():
    return redirect('/new')
```

Flow:

```text
User → /old
      ↓
Redirect
      ↓
/new
```

Common Uses:

- Renaming pages
- Login redirects
- Navigation workflows

# 13. Error Handling

Applications should fail gracefully.

---

## 404 Error

Page not found.

Example:

```python
@app.errorhandler(404)
def not_found(error):
    return "Page Not Found", 404
```

---

## 500 Error

Server error.

Example:

```python
@app.errorhandler(500)
def server_error(error):
    return "Internal Server Error", 500
```

Benefits:

- Better user experience
- Easier debugging
- Professional applications

---

## 500 Internal Server Error
