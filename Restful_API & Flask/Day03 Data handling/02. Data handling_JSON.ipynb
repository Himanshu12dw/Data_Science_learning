# What is JSON?

JSON stands for:

**JavaScript Object Notation**

It is a lightweight text-based format used for storing and exchanging data.

Although it originated from JavaScript, JSON is language-independent and supported by almost every programming language.

---

## Real Life Analogy

Imagine filling a form:

Name: Himanshu <br>
Age: 26 <br>
City: Delhi <br>

JSON stores this information as:

```
{
    "name": "Himanshu",
    "age": 26,
    "city": "Delhi"
}
```

Each piece of information is stored as:

Key : Value

This is called a Key-Value Pair.
___

___
# JSON Syntax Rules

JSON follows some important rules:

1. Data is stored as Key-Value pairs.
2. Keys must be enclosed in double quotes.
3. Strings must use double quotes.
4. Objects are enclosed using { }.
5. Arrays are enclosed using [ ].
6. Data is separated using commas.
7. JSON supports nested structures.

# Python Dictionary vs JSON

| Python Dictionary | JSON |
|------------|---------|
| Single Quotes Allowed | Double Quotes Required |
| True | true |
| False | false |
| None | null |
| dict | object |
| list | array |

Remember:

JSON and Python Dictionary look similar but are not exactly the same.

# JSON Module in Python

Python provides a built-in module called:

json

This module helps us:

1. Convert Python → JSON
2. Convert JSON → Python

# Convert Python Dictionary to JSON

Method:

json.dumps()

dumps = dump string

````python
import json

employee = {
    "name": "Himanshu",
    "age": 26
}

json_string = json.dumps(employee)

print(json_string)
print(type(json_string))
````

| Function       | Purpose              | Input         | Output         |
| -------------- | -------------------- | ------------- | -------------- |
| `json.dump()`  | Python → JSON File   | Python Object | Writes to file |
| `json.dumps()` | Python → JSON String | Python Object | JSON String    |
| `json.load()`  | JSON File → Python   | JSON File     | Python Object  |
| `json.loads()` | JSON String → Python | JSON String   | Python Object  |


````python
import json

employee = {
    "name": "Himanshu",
    "age": 26
}

json_string = json.dumps(employee)

print(json_string)
print(type(json_string))
````

````python
import json

json_string = '''
{
    "name":"Himanshu",
    "age":26
}
'''

python_dict = json.loads(json_string)

print(python_dict)
print(type(python_dict))
````

````python
import json

employee = {
    "name": "Himanshu",
    "age": 26
}

with open("employee.json", "w") as file:
    json.dump(employee, file, indent=4)

print(employee)
print(type(file))
````

````python
import json

with open("employee.json", "r") as file:
    data = json.load(file)

print(data)
print(type(data))
````

# Convert JSON String to Python Dictionary

Method:

json.loads()

loads = load string

````python
json_string = '''
{
    "name":"Himanshu",
    "age":26
}
'''

python_dict = json.loads(json_string)

print(type(json_string))
print(python_dict)
print(type(python_dict))
````

# JSON Files

Many APIs save responses into .json files.

Example:

employee.json

We can read and write JSON files easily.

````python
employee = {
    "name": "Himanshu",
    "age": 26,
    "city": "Delhi"
}

with open("employee.json", "w") as file:
    json.dump(employee, file, indent=4)

print("File Saved")

with open("employee.json", "r") as file:
    data = json.load(file)

print(data)
````

# Understanding Complex Nested API Responses

Real APIs are rarely flat.

````python
nested_data = {
    "user": {
        "id": 101,
        "name": "Himanshu",
        "address": {
            "city": "Delhi",
            "country": "India"
        }
    }
}

nested_data["user"]["address"]["city"]
````

# Working with Arrays in JSON

````python
employee = {
    "name": "Himanshu",
    "skills": [
        "Python",
        "SQL",
        "Power BI"
    ]
}

print(employee["skills"])
for skill in employee["skills"]:
    print(skill)
````

# Handling Missing Keys

Not every API response contains every key.

Using get() prevents errors.

````python
employee = {
    "name": "Himanshu"
}

# employee["salary"] #this will raise a KeyError since the key "salary" is not present in the dictionary

employee.get("salary")  # returns None if key is not present in dictionary

employee.get("salary", "Salary information not available")  #2nd argument is default value if key is not present in dictionary
````

# Common API Status Codes

| Code | Meaning |
|--------|---------|
| 200 | Success |
| 201 | Created |
| 400 | Bad Request |
| 401 | Unauthorized |
| 403 | Forbidden |

Always check status codes before processing JSON.

````python
response = requests.get("https://jsonplaceholder.typicode.com/users")

if response.status_code == 200:
    print("Success")
else:
    print("Failed")
````
