# What is Data Handling in APIs?

APIs exchange data between systems. A Data Scientist often consumes data from APIs and converts it into a format suitable for analysis or machine learning.

Common formats:
- JSON
- CSV
- XML
- Excel

---

# JSON (JavaScript Object Notation)

JSON is the most widely used format for APIs.

Example:

```json
{
  "name": "Himanshu",
  "age": 26,
  "skills": ["Python", "SQL"]
}
```

Advantages:
- Human-readable
- Lightweight
- Easy to parse
- Language independent

---

# requests Library

The requests library is the most popular Python library for working with APIs.

Install:

```bash
pip install requests
```

Features:
- GET
- POST
- PUT
- PATCH
- DELETE
- Headers
- Authentication
- Query Parameters
- JSON Parsing

Example:

```python
import requests

response = requests.get("https://api.example.com/users")

print(response.status_code)
print(response.json())
```

---

# GET Request

Used to retrieve data.

Example:

```python
requests.get(url)
```

Use Cases:
- Weather data
- Stock prices
- Customer records

---

# POST Request

Used to create resources.

```python
requests.post(url, json=data)
```

Example:
- User registration
- ML prediction requests

---

# PUT Request

Updates an entire resource.

```python
requests.put(url, json=data)
```

---

# PATCH Request

Updates only selected fields.

```python
requests.patch(url, json={"age":31})
```

---

# DELETE Request

Removes resources.

```python
requests.delete(url)
```

---

# Query Parameters

Parameters attached to URLs.

```python
params = {"city":"Delhi"}

requests.get(url, params=params)
```

Result:

```text
?city=Delhi
```

---

# json Library

Python's built-in JSON library.

```python
import json
```

Functions:
- loads()
- dumps()
- load()
- dump()

---

# json.loads()

Converts JSON string to Python object.

```python
json.loads('{"name":"John"}')
```

---

# json.dumps()

Converts Python object to JSON string.

```python
json.dumps({"name":"John"})
```

---

# json.load()

Reads JSON from a file.

```python
with open("data.json") as f:
    data = json.load(f)
```

---

# json.dump()

Writes JSON to a file.

```python
with open("data.json","w") as f:
    json.dump(data,f)
```

# CSV Files

CSV = Comma Separated Values

Example:

```text
name,age,salary
John,30,50000
```

Common in Data Science.

---

# pandas

Most important library for CSV processing.

Install:

```bash
pip install pandas
```

Read CSV:

```python
import pandas as pd

df = pd.read_csv("data.csv")
```

Write CSV:

```python
df.to_csv("output.csv")
```

---

# Handling Missing Values

```python
df.isnull().sum()
df.dropna()
df.fillna(0)
```

---

# Filtering Data

```python
df[df["age"] > 30]
```

---

# Grouping Data

```python
df.groupby("department").mean()
```

---

# csv Module

Built-in Python module.

```python
import csv

with open("data.csv") as f:
    reader = csv.reader(f)

    for row in reader:
        print(row)
```

Advantages:
- Lightweight
- Memory efficient

---

# pandas vs csv

| Feature | pandas | csv |
|----------|---------|---------|
| Analysis | Excellent | Manual |
| Cleaning | Easy | Manual |
| Large Files | Good | Good |

# NumPy

# XML

Extensible Markup Language.

Example:

```xml
<employee>
    <name>John</name>
</employee>
```

# ElementTree

Built-in XML library.

```python
import xml.etree.ElementTree as ET

tree = ET.parse("employees.xml")
root = tree.getroot()
```

---

# lxml
