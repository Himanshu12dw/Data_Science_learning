# Basics of Web Development


Web development is the process of building applications and services that run over the internet.

For Data Scientists, web development becomes important when a machine learning model needs to be used by real users. A model running on your laptop is useful for learning, but a model exposed through a web application or API can be used by customers, mobile apps, dashboards, and business systems.

Examples:
- House price prediction website
- Fraud detection service
- Recommendation system
- Customer churn prediction API


___
# Evolution of Communication Networks


## Telephonic Networks
Early communication systems were designed primarily for voice communication.

## ARPANET
In the 1960s, ARPANET introduced packet switching. Instead of sending a complete message at once, messages were divided into smaller packets that could travel independently and be reassembled later.

Benefits:
- Better reliability
- Efficient use of network resources
- Foundation of the modern internet

## Internet
The Internet is a global network connecting millions of computers and networks.

## World Wide Web (WWW)
Invented by Tim Berners-Lee in 1989.

The Internet is the infrastructure.
The World Wide Web is a service that runs on top of the Internet and allows users to access linked web pages through browsers.


___
# Client-Server Architecture


The web operates on a Client-Server model.

## Client
The system requesting information.
Examples:
- Chrome
- Firefox
- Mobile applications

## Server
The system providing information.
Examples:
- Google servers
- Netflix servers
- Banking servers

Flow:
Client -> Request -> Server
Server -> Response -> Client

Whenever you open a website, your browser acts as a client and communicates with a server.




___
# HTTP (HyperText Transfer Protocol)


HTTP is the communication protocol used between clients and servers.

A protocol is simply a set of rules.

When you type a website URL into a browser:
1. Browser sends an HTTP request.
2. Server processes the request.
3. Server returns an HTTP response.
4. Browser displays the result.

HTTP is stateless.
This means each request is independent and the server does not automatically remember previous requests.




___
# HTTP Request


An HTTP Request is a message sent from a client to a server.

Components:

## Request Line
Contains:
- HTTP Method
- URL
- HTTP Version

Example:
GET /users HTTP/1.1

## Headers
Metadata about the request.

Examples:
- Content-Type
- Authorization
- User-Agent

## Body
Optional section containing data.

Usually used with:
- POST
- PUT
- PATCH

Example JSON:

{
    "name": "John",
    "age": 30
}



___
# HTTP Response


An HTTP Response is sent by the server back to the client.

Components:

## Status Line
Contains:
- HTTP Version
- Status Code
- Status Message

Example:
HTTP/1.1 200 OK

## Headers
Additional information about the response.

Examples:
- Content-Type
- Content-Length

## Body
Contains the actual data returned by the server.

Example:

{
    "name": "John",
    "age": 30
}




___
# HTTP Methods - GET


GET is used to retrieve data from a server.

Characteristics:
- Read-only operation
- Does not modify data

Example:
GET /users

Use Cases:
- Loading a webpage
- Fetching customer information
- Retrieving records from a database


---
# HTTP Methods - POST


POST is used to create new resources.

Example:

POST /users

{
    "name": "John",
    "age": 30
}

Use Cases:
- User registration
- Creating orders
- Submitting forms
- most secure


___
# HTTP Methods - PUT


PUT is used to replace an entire resource.

Example:

PUT /users/1

{
    "name": "John",
    "age": 31
}

The complete resource is replaced with the new version.


___
# HTTP Methods - DELETE


DELETE removes a resource.

Example:

DELETE /users/1

Use Cases:
- Deleting users
- Removing products
- Deleting records


___
# HTTP Methods - PATCH


PATCH updates only selected fields.

Example:

PATCH /users/1

{
    "age": 32
}

Difference from PUT:
PUT replaces the whole resource.
PATCH updates only specific fields.


___
# HTTP Methods - HEAD


HEAD is similar to GET but returns only headers.

No response body is returned.

Uses:
- Checking if a resource exists
- Getting metadata
- Checking file size


___
# HTTP Methods - OPTIONS


OPTIONS tells a client what operations are allowed.

Example Response:
GET
POST
PUT
DELETE

Commonly used in browser-based applications and CORS checks.

# HTTP Methods Comparison

| Method | Purpose | CRUD Operation | Request Body | Changes Data? | Idempotent? | Typical Use Case | Example |
|----------|----------|----------|----------|----------|----------|----------|----------|
| GET | Retrieve data | Read | No | ❌ No | ✅ Yes | Fetch records | Get user details |
| POST | Create a new resource | Create | Yes | ✅ Yes | ❌ No | Create a user, submit form | Create new account |
| PUT | Replace an entire resource | Update | Yes | ✅ Yes | ✅ Yes | Update complete profile | Replace user information |
| PATCH | Partially update a resource | Update | Yes | ✅ Yes | Usually ✅ Yes | Update specific fields | Change email address |
| DELETE | Remove a resource | Delete | Usually No | ✅ Yes | ✅ Yes | Delete records | Delete user account |
| HEAD | Retrieve headers only | Read | No | ❌ No | ✅ Yes | Check metadata | Check file size |
| OPTIONS | Show supported methods | N/A | No | ❌ No | ✅ Yes | API capability discovery | Check allowed operations |

## Quick Summary

|----------|----------|

| POST | Create something new |
| PUT | Replace everything |

| DELETE | Remove something |

---

## PUT vs PATCH

### Existing Resource

```json
{
    "id": 1,
    "name": "John",
    "age": 30,
    "city": "Delhi"
}
```

### PUT Request

```json
{
    "name": "John",
    "age": 31,
    "city": "Mumbai"
}
```

Result:

```json
{
    "name": "John",
    "age": 31,
    "city": "Mumbai"
}
```

Entire resource is replaced.

### PATCH Request

```json
{
    "age": 31
}
```

Result:

```json
{
    "id": 1,
    "name": "John",
    "age": 31,
    "city": "Delhi"
}
```

Only specified fields are updated.

# HTTPS


HTTPS stands for HyperText Transfer Protocol Secure.

HTTPS protects communication between clients and servers.

Benefits:
1. Encryption
2. Authentication
3. Data Integrity

HTTPS uses TLS (Transport Layer Security).

When you see a lock icon in your browser, communication is encrypted and secured through HTTPS.



___
# **Role of APIs in Data Science**


API stands for Application Programming Interface.

An API allows different software systems to communicate.

Data Science applications heavily rely on APIs.

## Data Integration
APIs collect data from:
- Weather services
- Financial markets
- Social media
- External databases

## Real-Time Predictions
Users send data.
Machine learning models return predictions instantly.

Examples:
- Fraud detection
- Recommendation systems
- Dynamic pricing

## Model Deployment
APIs expose machine learning models as services.

Example:
/predict endpoint

Input:
{
  "area":1500,
  "bedrooms":3
}

Output:
{
  "price":8500000
}

## Automation
APIs automate:
- Data collection
- Data preprocessing
- Prediction pipelines

## Collaboration
Different teams can use the same API.
Frontend, backend, mobile, and data science teams can work independently.

# End-to-End Machine Learning API Flow


User-->Web Application-->API-->Machine Learning Model-->Prediction-->API Response-->Web Application-->User

This is how most modern AI products operate.


# Key Takeaways


1. The Internet connects computers globally.
2. The World Wide Web runs on top of the Internet.
3. Browsers and servers communicate using HTTP.
4. Requests are sent by clients and responses are returned by servers.
5. HTTP methods define operations on resources.
6. HTTPS secures communication.
7. APIs enable applications to communicate.
8. APIs are essential for deploying machine learning models.
9. Web development is a critical skill for Data Scientists moving toward production systems.
