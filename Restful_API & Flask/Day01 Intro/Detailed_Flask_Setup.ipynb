## What is Flask?
Flask is a lightweight Python web framework used to build websites, web applications, and APIs. It is especially popular among Data Scientists because it allows machine learning models to be exposed as web services.

# 3. Virtual Environment

A virtual environment isolates project dependencies.

Why needed?
- Prevents package conflicts
- Keeps projects independent
- Easier deployment

Commands:

```bash
pip install virtualenv

mkdir flask_project
cd flask_project

virtualenv venv
```

Activate:

Windows:

```bash
venv\Scripts\activate
```

Linux/Mac:

```bash
source venv/bin/activate
```

# 7. First Flask Application

app.py

```python
from flask import Flask

app = Flask(__name__)

@app.route('/')
def home():
    return "Hello Flask!"

if __name__ == "__main__":
    app.run(debug=True)
```

Run:

```bash
python app.py
```

Open:

http://127.0.0.1:5000

---

# 8. Understanding app.py

## Flask Application Object

```python
app = Flask(__name__)
```

Creates the application.

## Route

```python
@app.route('/')
```

Maps a URL to a function.

## View Function

```python
def home():
    return "Hello Flask!"
```

Executed when route is accessed.

---

# 11. Rendering HTML

```python
from flask import render_template

@app.route('/')
def home():
    return render_template('index.html')
```

Flask automatically looks inside templates folder.
